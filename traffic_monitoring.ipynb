```python
"""
Name: Sagar
Roll No: 2301010312
Course: Image Processing & Computer Vision
Unit: 4th
Assignment Title: Object Representation and Feature Extraction for Traffic Images
Date: 07-04-2026
"""

import cv2
import numpy as np
import os
import matplotlib.pyplot as plt

# Create output directory
OUTPUT_DIR = "outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# -----------------------------
# Display Function
# -----------------------------
def show(title, img):
    plt.figure()
    plt.title(title)
    if len(img.shape) == 2:
        plt.imshow(img, cmap='gray')
    else:
        plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.show()


# -----------------------------
# Task 1: Edge Detection
# -----------------------------
def edge_detection(image, name):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Sobel
    sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
    sobel = cv2.magnitude(sobelx, sobely)
    sobel = np.uint8(sobel)

    # Canny
    canny = cv2.Canny(gray, 100, 200)

    cv2.imwrite(f"{OUTPUT_DIR}/{name}_sobel.png", sobel)
    cv2.imwrite(f"{OUTPUT_DIR}/{name}_canny.png", canny)

    print("Sobel and Canny edge detection completed")

    show("Sobel Edge", sobel)
    show("Canny Edge", canny)

    return canny


# -----------------------------
# Task 2: Object Representation
# -----------------------------
def object_representation(edge_img, original, name):
    contours, _ = cv2.findContours(edge_img, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    output = original.copy()

    print("\nDetected Objects:")
    for i, cnt in enumerate(contours):
        area = cv2.contourArea(cnt)
        perimeter = cv2.arcLength(cnt, True)

        x, y, w, h = cv2.boundingRect(cnt)
        cv2.rectangle(output, (x, y), (x + w, y + h), (0, 255, 0), 2)

        print(f"Object {i+1}: Area={area:.2f}, Perimeter={perimeter:.2f}")

    cv2.imwrite(f"{OUTPUT_DIR}/{name}_contours.png", output)

    show("Contours & Bounding Boxes", output)


# -----------------------------
# Task 3: Feature Extraction
# -----------------------------
def feature_extraction(image, name):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # ORB
    orb = cv2.ORB_create()
    kp, des = orb.detectAndCompute(gray, None)

    output = cv2.drawKeypoints(image, kp, None, color=(0, 255, 0))

    cv2.imwrite(f"{OUTPUT_DIR}/{name}_orb.png", output)

    print(f"ORB detected {len(kp)} keypoints")

    show("ORB Keypoints", output)


# -----------------------------
# Task 4: Analysis
# -----------------------------
def analysis():
    print("\n--- Comparative Analysis ---")
    print("• Sobel detects gradient edges but is sensitive to noise.")
    print("• Canny provides cleaner and more accurate edges.")
    print("• ORB extracts important feature points useful for tracking.")
    print("• In traffic monitoring, features help detect vehicles, lanes, and motion.")


# -----------------------------
# Main Pipeline
# -----------------------------
def process_image(path, name):
    print(f"\nProcessing: {path}")

    image = cv2.imread(path)

    if image is None:
        print("Error loading image")
        return

    cv2.imwrite(f"{OUTPUT_DIR}/{name}_original.png", image)
    show("Original Image", image)

    # Task 1
    edges = edge_detection(image, name)

    # Task 2
    object_representation(edges, image, name)

    # Task 3
    feature_extraction(image, name)

    # Task 4
    analysis()


# -----------------------------
# Run for 3 Images
# -----------------------------
if __name__ == "__main__":
    print("Traffic Monitoring System")
    print("--------------------------")

    images = [
        ("Traffic1.jpg", "traffic1"),
        ("Traffic2.jpg", "traffic2"),
        ("Traffic3.jpg", "traffic3")
    ]

    for path, name in images:
        process_image(path, name)

    print("\nProcessing completed. Check outputs/ folder.")
```
