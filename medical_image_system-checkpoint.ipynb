```python
"""
Name: __________
Roll No: __________
Course: __________
Unit: __________
Assignment Title: Medical Image Processing System
Date: __________
"""

import cv2
import numpy as np
import os
import matplotlib.pyplot as plt

# Create output directory
OUTPUT_DIR = "outputs"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# -----------------------------
# Task 1: Run Length Encoding
# -----------------------------
def run_length_encoding(image):
    flat = image.flatten()
    encoding = []
    prev = flat[0]
    count = 1

    for pixel in flat[1:]:
        if pixel == prev:
            count += 1
        else:
            encoding.append((prev, count))
            prev = pixel
            count = 1
    encoding.append((prev, count))

    return encoding


def calculate_compression(original, encoded):
    original_size = original.size
    encoded_size = len(encoded) * 2  # (value, count)

    compression_ratio = original_size / encoded_size
    savings = (1 - (encoded_size / original_size)) * 100

    return compression_ratio, savings


# -----------------------------
# Task 2: Segmentation
# -----------------------------
def global_threshold(image, thresh=127):
    _, binary = cv2.threshold(image, thresh, 255, cv2.THRESH_BINARY)
    return binary


def otsu_threshold(image):
    _, binary = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    return binary


# -----------------------------
# Task 3: Morphological Ops
# -----------------------------
def morphological_processing(binary_img):
    kernel = np.ones((3, 3), np.uint8)
    dilation = cv2.dilate(binary_img, kernel, iterations=1)
    erosion = cv2.erode(binary_img, kernel, iterations=1)
    return dilation, erosion


# -----------------------------
# Display using matplotlib
# -----------------------------
def show(title, img):
    plt.figure()
    plt.title(title)
    plt.imshow(img, cmap='gray')
    plt.axis('off')
    plt.show()


# -----------------------------
# Processing Pipeline
# -----------------------------
def process_image(image_path, name_prefix):
    print(f"\nProcessing: {image_path}")

    image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

    if image is None:
        print(f"Error loading image: {image_path}")
        return

    # Save original
    cv2.imwrite(f"{OUTPUT_DIR}/{name_prefix}_original.png", image)

    # ---------------- Task 1 ----------------
    print("\n[Task 1] Image Compression (RLE)")
    rle = run_length_encoding(image)
    ratio, savings = calculate_compression(image, rle)

    print(f"Compression Ratio: {ratio:.2f}")
    print(f"Storage Savings: {savings:.2f}%")

    # Save RLE to file
    with open(f"{OUTPUT_DIR}/{name_prefix}_rle.txt", "w") as f:
        for val, count in rle[:1000]:
            f.write(f"{val}:{count} ")

    print("Displaying Original Image...")
    show("Original Image", image)

    # ---------------- Task 2 ----------------
    print("\n[Task 2] Image Segmentation")

    global_seg = global_threshold(image)
    otsu_seg = otsu_threshold(image)

    cv2.imwrite(f"{OUTPUT_DIR}/{name_prefix}_global.png", global_seg)
    cv2.imwrite(f"{OUTPUT_DIR}/{name_prefix}_otsu.png", otsu_seg)

    print("Displaying Global Threshold Image...")
    show("Global Threshold", global_seg)

    print("Displaying Otsu Threshold Image...")
    show("Otsu Threshold", otsu_seg)

    # ---------------- Task 3 ----------------
    print("\n[Task 3] Morphological Processing")

    dilated, eroded = morphological_processing(otsu_seg)

    cv2.imwrite(f"{OUTPUT_DIR}/{name_prefix}_dilation.png", dilated)
    cv2.imwrite(f"{OUTPUT_DIR}/{name_prefix}_erosion.png", eroded)

    print("Displaying Dilation Image...")
    show("Dilation", dilated)

    print("Displaying Erosion Image...")
    show("Erosion", eroded)

    # ---------------- Task 4 ----------------
    print("\n[Task 4] Analysis and Interpretation")
    print("• Global Thresholding: Uses fixed value, may miss fine details.")
    print("• Otsu's Method: Automatically selects optimal threshold.")
    print("• Dilation: Expands segmented regions (fills gaps).")
    print("• Erosion: Removes noise and shrinks regions.")
    print("• These operations help highlight important medical regions.")


# -----------------------------
# Main Execution (ONLY X-RAY)
# -----------------------------
if __name__ == "__main__":
    print("Medical Image Processing System")
    print("--------------------------------")

    process_image("23518-hand-x-ray.jpg", "xray")

    print("\nProcessing completed. Check outputs/ folder.")
```
